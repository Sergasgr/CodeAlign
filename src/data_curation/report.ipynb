```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import json

metadata = []
accepted_samples = 0
rejected_samples = 0
with open('./report_dataset.jsonl', 'r', encoding='utf-8') as r:
    for line in r:
        sample = json.loads(line)
        if sample["metadata"]["status"] == "accepted":
            accepted_samples += 1
        else:
            rejected_samples += 1
        metadata.append(dict(sample["metadata"]))

df = pd.DataFrame(metadata)

total_samples = len(df)
print(f"Samples analyzed: {total_samples}")
print(f"Samples accepted: {accepted_samples} ({(accepted_samples/total_samples):.2%})")
print(f"Samples discarded: {rejected_samples} ({(rejected_samples/total_samples):.2%})")
```

Pie Chart

```python
sizes = [accepted_samples, rejected_samples]
labels = ['Accepted', 'Rejected']
plt.pie(sizes, labels=labels, autopct='%1.1f%%', colors=['#4CAF50', '#F44336'])
plt.title("Pie Chart - Accepted vs Rejected Data")
plt.show()
```

Bar Chart 

```python
df_rejected = df[df["status"] != "accepted"]
sns.countplot(data=df_rejected, x='status')
plt.title("Failure Anlysis")
plt.show()
```
